# Проверка расчётов карманов ТМ
Оригинальные цены и исходы, один первый подходящий вход на матч и фильтр. Все результаты разведочные.

```python
import json, gzip
from pathlib import Path
from decimal import Decimal as D
p = Path.cwd()
rows = json.loads(gzip.decompress((p/'candidate-entries.json.gz').read_bytes()))
for r in rows:
    if r['final'] is None: continue
    t = sum(r['final']); q = round(r['line']*4); ticks = [q-1,q+1] if q%2 else [q]
    v = sum(D(str(r['price']))-1 if t*4<x else D(0) if t*4==x else D(-1) for x in ticks)/len(ticks)
    assert abs(float(v)-r['profit']) < 1e-9
print("Все сохранённые финансовые расчёты проверены")
```
